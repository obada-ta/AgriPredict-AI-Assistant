## 1.first of all , lets read the dataset:

```python
import numpy as  np
import pandas as pd

# display the data set to do changes on it :
df=pd.read_csv('C:\\Users\\husse\\Desktop\\Senior Project\\Datasets\\crop_yield_dataset.csv')
df.head()
```

## 2. we need to transform the data from strings to numbers:

```python
print(df['Crop_Type'].unique())
print(df['Soil_Type'].unique())
```

```python
from sklearn.preprocessing import LabelEncoder

# apply the LabelEncoder preprocessing on the <Crop_Type> column:
L1=LabelEncoder();
L1.fit(df[['Crop_Type']])
df['Crop_Type_New']=L1.transform(df[['Crop_Type']])

# apply the LabelEncoder preprocessing on the <Soil_Type> column:
L2=LabelEncoder();
L2.fit(df[['Soil_Type']])
df['Soil_Type_New']=L2.transform(df[['Soil_Type']])

#droping the columns that we did a <label encoding> on it :
df=df.drop('Crop_Type',axis = 1)
df=df.drop('Soil_Type',axis = 1)

# and drop the non_Used Column:
df=df.drop('Date',axis = 1)


df.head()
```

```python
# so far , we have done preprocessing on the data set by using <Label Encoder>,
# and we transformed all the data into numbers.
```

## 4. Start the machine learning and build the model process: 

```python
df.columns
```

```python
y = df[['Crop_Yield']]
X = df[['Soil_pH', 'Temperature', 'Humidity', 'Wind_Speed', 'N', 'P', 'K',
        'Soil_Quality', 'Crop_Type_New', 'Soil_Type_New']]
```

### 1. split the data into (train data) and (test data):

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(X,y,test_size = 0.3)
```

### 2- Apply Linear Regression on the data: 

```python
from sklearn.linear_model import LinearRegression

LR_model1 = LinearRegression()
LR_model1.fit(x_train, y_train)
print(LR_model1.score(x_train ,y_train))
print(LR_model1.score(x_test ,y_test))
```

### 3- Apply Random Forest on the data: 

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np
import pandas as pd

def evaluate_model(model, x_test, y_test, model_name):
    """
    Evaluates a regression model and prints performance metrics
    """
    y_pred = model.predict(x_test)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)
    
    print(f"\n{model_name} Performance:")
    print(f"MSE: {mse:.4f}")
    print(f"RMSE: {rmse:.4f}")
    print(f"R²: {r2:.4f}")
    
    return y_pred

# Random Forest Regressor
RF_model = RandomForestRegressor(
    n_estimators=100,      # Number of trees
    max_depth=10,
    min_samples_split=5,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1  # Use all available cores
)
RF_model.fit(x_train, y_train)

y_pred_rf = evaluate_model(RF_model, x_test, y_test, "Random Forest")

print('')
print(RF_model.score(x_train ,y_train))
print(RF_model.score(x_test ,y_test))
```
